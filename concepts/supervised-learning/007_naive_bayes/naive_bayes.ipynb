```python
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score
from sklearn.datasets import load_iris
import matplotlib.pyplot as plt
```

```python
data=load_iris()
x=data["data"]
y=data["target"]
train_x,test_x,train_y,test_y=train_test_split(x,y,random_state=50,test_size=0.2)
accuracy=[]
values = [1e-12, 1e-10, 1e-9, 1e-8, 1e-7,1e-1,0.3]
for i in values:
    model=GaussianNB(var_smoothing=i)
    model.fit(train_x,train_y)
    result=model.predict(test_x)
    score=accuracy_score(result,test_y)
    accuracy.append(score)
plt.plot(values, accuracy, marker="o")
plt.xscale("log")
plt.xlabel("var_smoothing")
plt.ylabel("Accuracy")
plt.title("Accuracy vs var_smoothing")
plt.show()
print(accuracy)
```

## Experiment: var_smoothing

The following values were tested:

| var_smoothing | Accuracy |
|---:|---:|
| 1e-12 | 93.33% |
| 1e-10 | 93.33% |
| 1e-9 | 93.33% |
| 1e-8 | 93.33% |
| 1e-7 | 93.33% |
| 1e-1 | 96.67% |
| 0.3 | 93.33% |

### Observation

Most of the tested values produced an accuracy of 93.33%.

Increasing `var_smoothing` to `0.1` improved the accuracy to 96.67%,
while increasing it further to `0.3` returned the accuracy to 93.33%.

Therefore, changing `var_smoothing` did affect the predictions in this
experiment, but the relationship was not simply that increasing the
parameter always improves or decreases accuracy.
