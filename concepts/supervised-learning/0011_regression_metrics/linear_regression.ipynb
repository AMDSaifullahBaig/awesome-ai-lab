```python
from sklearn.metrics import (r2_score,mean_squared_error,root_mean_squared_error,mean_absolute_error)
from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
```

```python
data=load_diabetes()
x=data.data
y=data.target
train_x,test_x,train_y,test_y=train_test_split(x,y,random_state=50,test_size=0.2)
model=LinearRegression()
model.fit(train_x,train_y)
result=model.predict(test_x)
print(f"The r2_score:   {r2_score(test_y,result)}")
print(f"The mean_squared_error:     {mean_squared_error(test_y,result)}")
print(f"The root_mean_squared_error:    {root_mean_squared_error(test_y,result)}")
print(f"The mean_absolute_error:    {mean_absolute_error(test_y,result)}")
```

MAE
→ Average absolute prediction error

MSE
→ Average squared prediction error

RMSE
→ Square root of MSE, in the target's units

R²
→ Proportion of target variation explained by the model
