```python
from sklearn.metrics import accuracy_score
from sklearn.model_selection import train_test_split
from sklearn.datasets import load_iris
from sklearn.ensemble import RandomForestClassifier
import matplotlib.pyplot as plt
```

```python
data=load_iris()
x=data["data"]
y=data["target"]
train_x,test_x,train_y,test_y=train_test_split(x,y,random_state=50,test_size=0.2)
accuracy=[]
for i in range(1,30):
    model=RandomForestClassifier(n_estimators=i,random_state=50)
    model.fit(train_x,train_y)
    result=model.predict(test_x)
    score=accuracy_score(test_y,result)
    accuracy.append(score)
plt.plot(range(1,30),accuracy)
plt.show()
for i,j in zip(data.feature_names,model.feature_importances_):
    print(f"{i}: {j} %")
print(accuracy)
```
