```python
from sklearn.metrics import accuracy_score
from sklearn.tree import DecisionTreeClassifier
import matplotlib.pyplot as plt
from sklearn.tree import plot_tree
from sklearn.model_selection import train_test_split
from sklearn.datasets import load_iris
```

```python
data=load_iris()
x=data["data"] 
y=data["target"]
train_x,test_x,train_y,test_y=train_test_split(x,y,random_state=50,test_size=0.2)
l=[]
for i in range(1,20):
    model=DecisionTreeClassifier(random_state=50,max_depth=i)
    model.fit(train_x,train_y)
    result=model.predict(test_x)
    score=accuracy_score(result,test_y)
    train_score=accuracy_score(test_y,result)
    l.append(score)
plt.plot(range(1,20), l, marker="x", label="Test")
plt.xlabel("Max Depth")
plt.ylabel("Accuracy")
plt.title("Decision Tree: Depth vs Accuracy")
plt.show()
print("Feature importance for decision tree:")
for i,j in zip(data.feature_names,model.feature_importances_):
    print(i,":",j*100,"%")
```
