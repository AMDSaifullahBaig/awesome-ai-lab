```python
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
import matplotlib.pyplot as plt
```

```python
data=load_iris()
x=data["data"]
y=data["target"]
train_x,test_x,train_y,test_y=train_test_split(x,y,random_state=50,test_size=0.2)
accuracy=[]
lr= [0.01, 0.05, 0.1, 0.2]
est= [10, 50, 100, 200]
for i,j in zip(lr,est):
    model=GradientBoostingClassifier(learning_rate=i,n_estimators=j,random_state=50)
    model.fit(train_x,train_y)
    result=model.predict(test_x)
    score=accuracy_score(result,test_y)
    accuracy.append(score)
plt.plot(lr,accuracy)
plt.title("accuracy vs lr")
plt.ylabel("accuracy")
plt.xlabel("learning rate")
plt.show()
print(max(accuracy))
```

### Observation

Different combinations of `learning_rate` and `n_estimators` were tested.
The tested configurations produced the same test accuracy of 96.67%.

This shows that changing these hyperparameters does not necessarily
change the final test accuracy, especially when evaluating on a small
test set.

The models may still differ internally even when their final accuracy
is identical.
