```python
import torch
torch.manual_seed(50)
x=torch.rand(1,5)
actual=torch.rand(1,5)
w=torch.rand(1,5,requires_grad=True)
b=torch.rand(1,5,requires_grad=True)
epoch=150
for i in range(epoch):
    y=x*w+b
    y=torch.relu(y)
    #y=torch.sigmoid(y)
    #y=torch.tanh(y)
    loss=torch.mean((y-actual)**2)
    lr=0.1
    loss.backward()
    with torch.no_grad():
        w-=lr*w.grad
        b-=lr*b.grad
    w.grad.zero_()
    b.grad.zero_()
    if i%30==0:print(loss.item())
```
