```python
import torch
no_features=torch.tensor(5)
k=torch.rand(5,5)
q=torch.rand(5,5)
v=torch.rand(5,5)
similarity=k@q.T
scaled=similarity/torch.sqrt(no_features)
sigmoid_scaled=torch.softmax(scaled,dim=1)
attention=sigmoid_scaled@v


print(f"similarity: {similarity}\n")
print(f"scaled similarity: {scaled}\n")
print(f"softmax of the scaled similarity: {sigmoid_scaled}\n")
print(f"attention score: {attention}\n")
```

## Attention Score

The attention score measures how strongly each **Query (Q)** relates to each **Key (K)**.

### 1. Dot Product

$$
\text{Score} = QK^T
$$

### 2. Scaling

Large scores can make Softmax too extreme, so we divide by $\sqrt{d_k}$:

$$
\text{Scaled Score} = \frac{QK^T}{\sqrt{d_k}}
$$

where $d_k$ is the dimension of the Key vectors.

### 3. Softmax

Convert scores into attention weights:

$$
\text{Attention Weights}
=
\text{softmax}
\left(
\frac{QK^T}{\sqrt{d_k}}
\right)
$$

### 4. Apply to Values

The final attention output is:

$$
\text{Attention}(Q,K,V)
=
\text{softmax}
\left(
\frac{QK^T}{\sqrt{d_k}}
\right)V
$$

**In short:**
`QKᵀ` → measures relevance
`Scaling` → keeps values stable
`Softmax` → converts relevance into weights
`× V` → combines the relevant information
