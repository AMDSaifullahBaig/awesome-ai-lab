```python
import torch
from torch import nn

x=torch.rand(1,1,1,1)
conv=nn.Conv2d(in_channels=1,out_channels=2,kernel_size=3,stride=2,padding=2)
conv(x).shape
```

| parameters | meaning |
|-|-|
| in_channels   | input filters |
| out_channels  | output filters |
| kernel_size   | sliding window block size |
| stride        | number of steps |
| padding       | adding empty info to the input to preserve dimention |

$$Outputdim = \left\lfloor \frac{Inputdim - Kernel + 2Padding}{stride} \right\rfloor + 1$$
