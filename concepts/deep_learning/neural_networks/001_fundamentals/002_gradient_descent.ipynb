### Overview

<h3><center> Gradient Descent  - torch.optim.SGD  </center></h3>

````python
import torch
torch.manual_seed(50)
x=torch.rand(10,5)
actual_result=torch.rand(10,10)
w=torch.rand(5,10,requires_grad=True)
b=torch.rand(1,10,requires_grad=True)
optimizer=torch.optim.SGD([w,b],lr=0.1)
epoch=200
for i in range(epoch+1):
    predicted_result=x@w+b
    loss=torch.mean((predicted_result-actual_result)**2)
    loss.backward()
    optimizer.step()
    optimizer.zero_grad()
    if i%50==0:
        print(f"epoch {i}:    {loss}")
````
