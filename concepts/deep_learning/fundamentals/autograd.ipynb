```python
import torch
x=torch.tensor([2.0],dtype=float,requires_grad=True)
y=x**2+2*x**3
y.backward()
print(x.grad)
```

```python
x=torch.tensor([14.0])
w=torch.tensor([5.0],requires_grad=True)
b=torch.tensor([2.0],requires_grad=True)
y=w*x+b
target=torch.tensor([10])
loss=(y-target)**2
loss.backward()
print(w.grad)
print(b.grad)
lr = 0.001
w = w - lr * w.grad
b = b - lr * b.grad
y=w*x+b
print("before grad:",loss)
loss=(y-target)**2
print("after grad:",loss)
```

```python
x=torch.tensor([14.0])
w=torch.tensor([5.0],requires_grad=True)
b=torch.tensor([2.0],requires_grad=True)
target=torch.tensor([10])

for i in range(50):
    y=x*w+b
    loss=(y-target)**2
    loss.backward()
    lr=0.001
    with torch.no_grad():
        w-=lr*w.grad
        b-=lr*b.grad
    w.grad.zero_()
    b.grad.zero_()
    print(f"epoch {i}:{loss.item()}")
```
