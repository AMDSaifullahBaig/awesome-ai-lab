```python
from sklearn.datasets import load_iris
from sklearn.cluster import KMeans
```

```python
data=load_iris()
x=data.data
y=data.target
inertia=[]
for i in range(1,15):
    model=KMeans(n_clusters=i,random_state=50)
    model.fit(x)
    inertia.append(model.inertia_)
l=model.labels_
```

```python
import matplotlib.pyplot as plt
plt.plot(range(1,15), inertia, marker="o")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")
plt.title("Elbow Method")
plt.show()
```

## Elbow Method

The Elbow Method helps estimate a suitable number of clusters.

K-Means provides `inertia_`, which measures the sum of squared distances
between each data point and its assigned cluster centroid.

As the number of clusters increases, inertia decreases. However, after
a certain point, adding more clusters produces only a small improvement.

The point where this improvement begins to diminish is called the
"elbow".

For the Iris dataset, the elbow appears around K = 3.

```python
model=KMeans(n_clusters=3,random_state=50)
model.fit(x)
plt.scatter(x[:, 0], x[:, 1], c=model.labels_)

plt.scatter(
    model.cluster_centers_[:, 0],
    model.cluster_centers_[:, 1],
    marker="X",
    s=200
)

plt.xlabel("Sepal Length")
plt.ylabel("Sepal Width")
plt.title("K-Means Clusters")
plt.show()
```
